```python
import seaborn as sns
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

sns.set_theme(style="darkgrid")
```

```python
df = pd.read_csv("../data-pipeline/datasets/gold-data/untitledScore_Gold10k.csv")
df = df[df['score'] != 'unknown']
df = df[df['price'] != 'unknown']
df['score'] = df['score'].str.replace(';', '')
df = df.astype({'score': np.float16, 'price': np.float16})
df['score'] = df['score'].astype(np.int8)
```

# Plot showing the correlation between the price and review score of items

```python
score_plot = sns.stripplot(data=df, x='score', y='price', jitter=True, alpha=0.3)

score_plot.set_xlabel("Score out of 5")
score_plot.set_ylabel("Price in USD")
score_plot.set_yticks(range(0, 100, 10))
plt.show()
```

We can conclude a few things from this graph:
1. Users are less likely to rate items 2 or 3 when the scale is 1-5
2. It is more likely for users to give products a 5/5 rating
3. Products that cost $20+ get more 4/5 and 5/5 ratings

We must understand that this shows off user behavior more than it does show the quality of the products.\
Users would give perfect ratings to items they're simply satisfied with; they're not necessarly perfect products, this phenomenon is what makes solely relying on ratings inadaquate.\
